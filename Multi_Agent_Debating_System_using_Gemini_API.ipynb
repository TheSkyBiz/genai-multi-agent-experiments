```python
from google.colab import userdata
api_key = userdata.get('ga')

import os
import google.generativeai as genai

os.environ["GEMINI_API_KEY"] = api_key
genai.configure(api_key=os.environ["GEMINI_API_KEY"])

MODEL_NAME = "gemini-2.0-flash"
model = genai.GenerativeModel(MODEL_NAME)

# Utility function for nicer output
def pretty_print(text: str):
    display(Markdown(text))
```

```python
PRO_PROMPT = """You are Pro-Agent.
Argue strongly IN FAVOR of the given topic.
Keep your tone persuasive, logical, and structured.
Keep your response short and crisp, between 120–150 words.
Avoid long essays. Be clear.
Respond directly to the opponent when needed."""

CON_PROMPT = """You are Con-Agent.
Argue strongly AGAINST the given topic.
Keep your tone persuasive, logical, and structured.
Keep your response short and crisp, between 120–150 words.
Avoid long essays. Be clear.
Respond directly to the opponent when needed."""

JUDGE_PROMPT = """You are Judge-Agent.
Read the debate transcript (Pro vs Con).
Evaluate based on clarity, logic, persuasiveness.
Finally, declare the WINNER with a short explanation.
Give a concise judgement in 40–60 words."""
```

```python
def pro_agent(topic, history):
    prompt = f"{PRO_PROMPT}\n\nDebate topic: {topic}\nConversation so far:\n{history}\n\nPro-Agent:"
    return model.generate_content(prompt, generation_config={
        "max_output_tokens": 150,
        "temperature": 0.2
    }).text.strip()

def con_agent(topic, history):
    prompt = f"{CON_PROMPT}\n\nDebate topic: {topic}\nConversation so far:\n{history}\n\nCon-Agent:"
    return model.generate_content(prompt, generation_config={
        "max_output_tokens": 150,
        "temperature": 0.2
    }).text.strip()

def judge_agent(topic, history):
    prompt = f"{JUDGE_PROMPT}\n\nDebate topic: {topic}\nTranscript:\n{history}\n\nJudge:"
    return model.generate_content(prompt, generation_config={
        "max_output_tokens": 80,
        "temperature": 0.1
    }).text.strip()
```

```python
def run_debate(topic, rounds):
    history = ""
    for r in range(1, rounds+1):
        display(Markdown(f"### --- ROUND {r} ---"))

        # Pro-Agent
        pro_text = pro_agent(topic, history)
        display(Markdown(f"**Pro-Agent:**\n\n> {pro_text}"))
        history += f"\nPro-Agent: {pro_text}"

        # Con-Agent
        con_text = con_agent(topic, history)
        display(Markdown(f"**Con-Agent:**\n\n> {con_text}"))
        history += f"\nCon-Agent: {con_text}"

    # Judge Decision
    display(Markdown("## --- JUDGE DECISION ---"))
    judgement = judge_agent(topic, history)
    display(Markdown(judgement))
```

```python
run_debate("Should AI be used in education?", rounds=3)
```
