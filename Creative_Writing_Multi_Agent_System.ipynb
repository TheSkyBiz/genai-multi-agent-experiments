```python
from google.colab import userdata
api_key = userdata.get('ga')

import os
import google.generativeai as genai

os.environ["GEMINI_API_KEY"] = api_key
genai.configure(api_key=os.environ["GEMINI_API_KEY"])

model = "gemini-2.0-flash"

def run_gemini(prompt, model= model, temperature=0.7):
    model_obj = genai.GenerativeModel(model)
    response = model_obj.generate_content(prompt, generation_config={"temperature": temperature})
    return response.text.strip()

# Utility function for nicer output
def pretty_print(text: str):
    display(Markdown(text))
```

```python
def brainstormer_agent(topic):
    prompt = f"""
    You are the Brainstormer Agent.
    Generate a list of 5-7 creative ideas, outlines, or directions for the topic: "{topic}".
    Keep it concise, structured, and inspiring.
    """
    return run_gemini(prompt)
```

```python
def drafting_agent(outline, topic):
    prompt = f"""
    You are the Drafting Agent.
    Expand the following outline into a structured first draft article on "{topic}".
    Make it clear, engaging, and coherent.

    Outline:
    {outline}
    """
    return run_gemini(prompt)
```

```python
def editor_agent(draft):
    prompt = f"""
    You are the Editor Agent.
    Edit the following draft to improve grammar, flow, and clarity.
    Ensure it reads smoothly and has a professional tone.

    Draft:
    {draft}
    """
    return run_gemini(prompt)
```

```python
def fact_checker_agent(draft):
    prompt = f"""
    You are the Fact-Checker Agent.
    Review the following draft for factual accuracy.
    Identify any claims that may be questionable or need citations.
    Provide a list of flagged points and suggested corrections.

    Draft:
    {draft}
    """
    return run_gemini(prompt)
```

```python
def creative_ai_writing_assistant(topic):
    ideas = brainstormer_agent(topic)
    draft = drafting_agent(ideas, topic)
    edited = editor_agent(draft)
    fact_check = fact_checker_agent(edited)

    return edited, fact_check

# Run interactively
user_topic = input("Enter your writing topic: ")
final_draft, fact_check_report = creative_ai_writing_assistant(user_topic)

print("\n✅ Final Polished Draft:\n", final_draft)
print("\n🔍 Fact-Check Report:\n", fact_check_report)
```
